```python
import pandas as pd
from sklearn.preprocessing import StandardScaler
from sklearn.cluster import KMeans
```

```python
path = "../data/bronze/bus.xlsx"
```

```python
district = pd.read_excel(path,sheet_name=1)
```

```python
district= district[district["0a. Has committed ESBs?"]== 'yes']
```

```python
df_filtre = district[['3i. Percent of fleet that is electric',
                      '4f. Median household income',
                      '4g. Percent of population below the poverty level'
                      ]]
```

```python
df_filtre.isna().mean() * 100
```

```python
df_propre = df_filtre.dropna()
```

```python
df_propre.head(5)
```

```python
df_propre.corr()
```

```python
cols = [
            '3i. Percent of fleet that is electric',
            '4f. Median household income',
            '4g. Percent of population below the poverty level'
]

for col in cols:
    df_filtre[col] = df_filtre[col].fillna(df_filtre[col].median())
```

```python
df_propre.isna().mean() * 100
```

```python
df_propre.head(10)
```

```python
features_stdr = [
            '3i. Percent of fleet that is electric',
            '4f. Median household income',
            '4g. Percent of population below the poverty level'
]
```

```python
scaler = StandardScaler()

df_propre[features_stdr] = scaler.fit_transform(df_propre[features_stdr]) 
```

```python
df_propre.to_csv('../TP5/anomalies.csv',index=False)
```
